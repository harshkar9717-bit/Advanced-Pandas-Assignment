Q2 Sales Filtering Analysis

Filter the dataset to display:

Orders where Sales > 3500

Orders where Region = "East" and Sales > 2500

Explain what insights you observe

```python
# 1. Orders where Sales > 3500
high_sales = df[df['Sales'] > 3500]
print("--- Orders with Sales > 3500 ---")
print(high_sales)

# 2. Orders where Region = "East" and Sales > 2500
east_high_sales = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]
print("\n--- East Region Orders with Sales > 2500 ---")
print(east_high_sales)
```

Observed Insights: This analysis isolates high-ticket transactions. It highlights top-performing products/categories and identifies key revenue-driving deals in the East region, helping sales teams pinpoint high-value target accounts.

Q3 Customer Purchase Analysis

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.

```python
# Group by customer and sum sales, then sort descending and get top 5
top_customers = df.groupby('Customer')['Sales'].sum().reset_index()
top_customers = top_customers.sort_values(by='Sales', ascending=False).head(5)

print("--- Top 5 Customers by Total Sales ---")
print(top_customers)
```

Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?

```python
# Calculate total sales per region
regional_sales = df.groupby('Region')['Sales'].sum().reset_index()
# Sort to easily identify the top region
regional_sales = regional_sales.sort_values(by='Sales', ascending=False)

print("--- Regional Sales Performance ---")
print(regional_sales)

highest_region = regional_sales.iloc[0]['Region']
highest_revenue = regional_sales.iloc[0]['Sales']
print(f"\n⭐ Highest Revenue Generator: {highest_region} region with total sales of {highest_revenue:.2f}")
```

Q5 Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?

```python
# Calculate total quantity sold for each product
product_popularity = df.groupby('Product')['Quantity'].sum().reset_index()
product_popularity = product_popularity.sort_values(by='Quantity', ascending=False)

print("--- Product Popularity (Quantity Sold) ---")
print(product_popularity)

most_popular_product = product_popularity.iloc[0]['Product']
print(f"\n⭐ Most Popular Product: {most_popular_product}")
```

Q6 Multi-Level Grouping

Group the dataset by: Region and Category

Calculate the total sales for each combination.

What insights can you derive from this analysis?

```python
# Multi-level grouping
multi_group = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()

print("--- Total Sales by Region and Category ---")
print(multi_group)
```

Observed Insights: This breakdown reveals exactly which types of products thrive in specific geographic zones. For example, if "Technology" sales dominate the East while "Furniture" rules the West, inventory and marketing strategies can be localized for higher conversion.

Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?

```python
# Create the pivot table
sales_pivot = df.pivot_table(values='Sales', index='Region', columns='Product', aggfunc='sum', fill_value=0)

print("--- Regional Product Sales Pivot Table ---")
print(sales_pivot)

# Find which region sells the most laptops (Assuming your column is named 'Laptop')
if 'Laptop' in sales_pivot.columns:
    best_laptop_region = sales_pivot['Laptop'].idxmax()
    print(f"\n⭐ The region that sells the most laptops is: {best_laptop_region}")
else:
    print("\n⚠️ 'Laptop' column not found in the product list. Double-check your exact product names.")
```

Q8 Profitability Analysis

Calculate the average profit for each product category.

Which category is most profitable?

```python
# Calculate average profit for each product category
category_profit = df.groupby('Category')['Profit'].mean().reset_index()
category_profit = category_profit.sort_values(by='Profit', ascending=False)

print("--- Average Profit by Category ---")
print(category_profit)

most_profitable = category_profit.iloc[0]['Category']
print(f"\n⭐ Most Profitable Category: {most_profitable}")
```

Q9 Discount Impact Study

Analyze the relationship between discount and profit.

Tasks:

Calculate the average profit for each discount level

Identify whether higher discounts lead to lower profit

```python
# Calculate average profit for each discount level
discount_impact = df.groupby('Discount')['Profit'].mean().reset_index()

print("--- Discount Level vs Average Profit ---")
print(discount_impact)
```

Observed Insights: Generally, tracking this data reveals a negative correlation: as discount percentages step up, profit margins compress. This study helps businesses find the "sweet spot" discount threshold where conversion increases without destroying profit margins.

Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# Step 1: Filter the dataset
filtered_df = df[df['Sales'] > 3000]

# Step 2 & 3: Group data and calculate total sales
grouped_df = filtered_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

# Step 4: Reshape the result into a pivot table
wrangled_pivot = grouped_df.pivot_table(values='Sales', index='Region', columns='Product', fill_value=0)

print("--- Final Wrangled Data Pivot Table ---")
print(wrangled_pivot)
```

Business Insights Generated   High-Value Item Distribution: This pivot table maps out where premium transactions ({Sales} > \$3000$) are occurring. If a region is entirely missing from this table, it suggests that market only buys budget items.  Resource Allocation: Executive teams can use this map to deploy targeted corporate sales teams and high-end inventory specifically to the regions showing high-ticket momentum.
